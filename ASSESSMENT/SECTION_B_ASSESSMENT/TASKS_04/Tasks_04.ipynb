## Complete Python Script

```python
import pandas as pd

# ========================================
# FOOD DELIVERY PERFORMANCE EDA
# ========================================

# Load the CSV exported from Excel
df = pd.read_csv("food_delivery_orders.csv")

print("\n========================================")
print("FOOD DELIVERY PERFORMANCE EDA")
print("========================================")

# ========================================
# 1. DESCRIPTIVE STATISTICS
# ========================================

print("\nDELIVERY TIME DESCRIPTIVE STATISTICS")
print("----------------------------------------")

print(df["DeliveryTime"].describe())

# ========================================
# 2. MANUAL IQR CALCULATION
# ========================================

Q1 = df["DeliveryTime"].quantile(0.25)
Q3 = df["DeliveryTime"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nIQR ANALYSIS")
print("----------------------------------------")
print(f"Q1 (25th Percentile): {Q1:.2f}")
print(f"Q3 (75th Percentile): {Q3:.2f}")
print(f"IQR:                  {IQR:.2f}")
print(f"Lower Bound:          {lower_bound:.2f}")
print(f"Upper Bound:          {upper_bound:.2f}")

# ========================================
# 3. OUTLIER DETECTION
# ========================================

df["Outlier Flag"] = (
    (df["DeliveryTime"] < lower_bound) |
    (df["DeliveryTime"] > upper_bound)
)

outliers = df[df["Outlier Flag"]]

print("\nPOTENTIAL OUTLIERS")
print("----------------------------------------")

print(
    outliers[
        ["OrderID", "DeliveryTime", "Rating", "Revenue"]
    ]
)

print(f"\nTotal Potential Outliers: {len(outliers)}")

# ========================================
# 4. CORRELATION ANALYSIS
# ========================================

correlation_matrix = df[
    ["Revenue", "DeliveryTime", "Rating"]
].corr()

print("\nCORRELATION MATRIX")
print("----------------------------------------")
print(correlation_matrix)

# Business interpretation comments:
# Negative DeliveryTime-Rating correlation means that longer delivery
# times are associated with lower customer ratings in this dataset.

# Positive Revenue-DeliveryTime correlation means that higher-revenue
# orders tend to be associated with longer delivery times in this dataset.

# A correlation close to zero between Revenue and Rating means there is
# little linear relationship between revenue and customer rating.

# Correlation shows association, not causation.
```
